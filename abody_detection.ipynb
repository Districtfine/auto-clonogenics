# Nucleolar A-body detection (Congo red)

Rough approach for detecting nucleolar amyloid bodies (A-bodies, Audas et al.) in the `.czi` scans under `../CD7*`. Different problem shape from the colony counter:

- Source: 2-channel Zeiss `.czi` (Congo red + Hoechst), one FOV per condition — no multi-well grid, so no FastSAM well-detection step needed here.
- Reading `.czi`: need a new dependency (`czifile` for simple single-scene files, or `aicsimageio` + `aicspylibczi` if these turn out to be tiled/z-stacked). Not yet added to `pyproject.toml`.
- Segmentation target: A-bodies are small, punctate, variable-morphology structures — a stock Cellpose model likely won't segment them well out of the box.
- Planned approach: **human-in-the-loop** with Cellpose.
  1. Run a generic Cellpose model on a handful of Congo red crops.
  2. Manually correct the masks in the Cellpose GUI (~20-50 images).
  3. Retrain a custom Cellpose model from those corrections.
  4. Repeat 2-3 rounds until segmentation is good enough to trust unsupervised.
  - Hoechst channel can help restrict search to in/near-nucleolus regions if needed.
- Same device-portability note as the colony counter notebook: keep `mps`/`cuda`/`cpu` picked automatically, not hardcoded, since this also needs to run on the 1650 Ti box eventually.

Nothing implemented yet — parking this here while we focus back on the colony counter.
