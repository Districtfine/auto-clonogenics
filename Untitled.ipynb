# Colony counting pipeline

Isolate each well from a multi-well plate `.tif` scan, then run each well crop through Cellpose to count colonies and estimate diameter. Source scans live in `../Clonogenics`.

```python
import os
import warnings
warnings.filterwarnings("ignore", message="Sparse invariant checks")

import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tifffile as tifi
import torch
from ultralytics import FastSAM
from cellpose import models

# Make our inline plots look nice and big
plt.rcParams['figure.figsize'] = [12, 6]
```

## 1. Pick device & load models

Picks `mps`/`cuda`/`cpu` automatically (portable between Apple Silicon and the 1650 Ti box), reads the target plate scan, loads FastSAM (well extractor) and Cellpose (colony counter).

```python
# Point this directly to your test file
input_path = "../Clonogenics/4h HS 43 + radiation001.tif"

# Portable device pick: mps on Apple Silicon, cuda on the 1650 Ti box, cpu fallback.
# DEVICE (str) feeds ultralytics/FastSAM; TORCH_DEVICE (torch.device) feeds Cellpose.
if torch.backends.mps.is_available():
    DEVICE = "mps"
elif torch.cuda.is_available():
    DEVICE = "cuda"
else:
    DEVICE = "cpu"
TORCH_DEVICE = torch.device(DEVICE)
print(f"Using device: {DEVICE}")

print(f"Reading {input_path}...")
img = tifi.imread(input_path)
img_rgb = cv2.cvtColor(img, cv2.COLOR_GRAY2RGB) if len(img.shape) == 2 else img[..., :3] 

print("Loading FastSAM (Well Extractor) into GPU...")
sam_model = FastSAM('FastSAM-s.pt')

print("Loading Cellpose (Colony Counter) into GPU...")
cp_model = models.CellposeModel(gpu=DEVICE != "cpu", device=TORCH_DEVICE, model_type='cpsam_v2')
print("Done!")
```

## 2. Detect wells & build grid

Runs FastSAM on the plate scan, filters detections by circularity/area to find stained wells, then snaps them onto a consensus 2-plate × (2 col × 3 row) grid so missed/occluded wells still get an inferred position. Produces `valid_wells`: `(x, y, r)` per well.

```python
X_LIMIT_FRAC = 0.60
MIN_SPACING  = 150
N_COLS = 2
N_ROWS = 6
MIN_WELL_AREA = img_rgb.shape[0] * img_rgb.shape[1] * 0.008  # well ~ 0.8%+ of scan area

x_cutoff = img_rgb.shape[1] * X_LIMIT_FRAC

results = sam_model.predict(source=img_rgb, device=DEVICE, imgsz=1024, conf=0.25, iou=0.5, verbose=False)
masks = results[0].masks

# --- 1. Detect stained wells ---
detected = []
for mask in masks.xy:
    cnt = np.array(mask, dtype=np.int32)
    area = cv2.contourArea(cnt)
    perimeter = cv2.arcLength(cnt, True)
    circularity = 4*np.pi*area/(perimeter**2) if perimeter > 0 else 0
    if area > MIN_WELL_AREA and circularity > 0.7:
        M = cv2.moments(cnt)
        if M["m00"] != 0:
            cx, cy = int(M["m10"]/M["m00"]), int(M["m01"]/M["m00"])
            if cx <= x_cutoff and all((cx-ox)**2+(cy-oy)**2 >= MIN_SPACING**2
                                      for ox, oy in detected):
                detected.append((cx, cy))

# --- 2. Split detections into the two physical plates, then grid each one ---
def split_into_k(values, k):
    vals = sorted(values)
    if len(vals) <= k:
        return [int(v) for v in vals]
    gaps = sorted(((vals[i+1]-vals[i], i) for i in range(len(vals)-1)), reverse=True)
    cut_after = sorted(i for _, i in gaps[:k-1])
    groups, start = [], 0
    for c in cut_after:
        groups.append(vals[start:c+1]); start = c+1
    groups.append(vals[start:])
    return [int(np.mean(g)) for g in groups]

def group_by_gaps(points, axis, k):
    """Partition points into k groups along an axis (0=x, 1=y) at the k-1 largest gaps."""
    pts = sorted(points, key=lambda p: p[axis])
    if len(pts) <= k:
        return [[p] for p in pts]
    coords = [p[axis] for p in pts]
    gaps = sorted(((coords[i+1]-coords[i], i) for i in range(len(coords)-1)), reverse=True)
    cut_after = sorted(i for _, i in gaps[:k-1])
    groups, start = [], 0
    for c in cut_after:
        groups.append(pts[start:c+1]); start = c+1
    groups.append(pts[start:])
    return groups

N_COLS, N_ROWS_PER_PLATE, N_PLATES = 2, 3, 2  # two stacked 6-well plates

# radius from the tightest column spacing (robust to the big inter-plate gap)
all_col_xs = split_into_k([p[0] for p in detected], N_COLS)
col_gap = np.mean(np.diff(sorted(all_col_xs))) if len(all_col_xs) > 1 else 0

# split detections into the two plates by the large vertical gap between them
plates = group_by_gaps(detected, axis=1, k=N_PLATES)

final = []
for plate_pts in plates:
    col_xs = split_into_k([p[0] for p in plate_pts], N_COLS)
    row_ys = split_into_k([p[1] for p in plate_pts], N_ROWS_PER_PLATE)

    row_gap = np.mean(np.diff(sorted(row_ys))) if len(row_ys) > 1 else col_gap
    well_r  = int(min(g for g in (col_gap, row_gap) if g > 0) / 2)

    # assign each detected well in THIS plate to its nearest (row, col) cell
    assign = {}
    for (px, py) in plate_pts:
        c = min(range(len(col_xs)), key=lambda c: abs(px - col_xs[c]))
        r = min(range(len(row_ys)), key=lambda r: abs(py - row_ys[r]))
        assign[(r, c)] = (px, py)

    # consensus grid lines from the detections (median => robust to one bad centroid)
    def col_consensus(c):
        xs = [pos[0] for (rr, cc), pos in assign.items() if cc == c]
        return np.median(xs) if xs else col_xs[c]
    def row_consensus(r):
        ys = [pos[1] for (rr, cc), pos in assign.items() if rr == r]
        return np.median(ys) if ys else row_ys[r]

    tol_x = 0.4 * col_gap          # a detection this far off its column is an outlier
    tol_y = 0.4 * row_gap

    for r in range(N_ROWS_PER_PLATE):
        for c in range(N_COLS):
            exp_x, exp_y = int(col_consensus(c)), int(row_consensus(r))
            if (r, c) in assign:
                px, py = assign[(r, c)]
                x = exp_x if abs(px - exp_x) > tol_x else px   # snap only gross outliers
                y = exp_y if abs(py - exp_y) > tol_y else py
                snapped = (x != px) or (y != py)
                final.append((x, y, well_r, snapped))          # snapped => flagged like inferred
            else:
                final.append((exp_x, exp_y, well_r, True))

final.sort(key=lambda p: (p[1], p[0]))
print(f"Final Count: {len(final)} wells "
      f"({sum(not f[3] for f in final)} detected, {sum(f[3] for f in final)} inferred/snapped).")

final.sort(key=lambda p: (p[1], p[0]))
print(f"Final Count: {len(final)} wells "
      f"({sum(not f[3] for f in final)} detected, {sum(f[3] for f in final)} inferred).")

# --- 4. Preview ---
fig, ax = plt.subplots(figsize=(10, 10))
ax.imshow(img_rgb)
ax.axvline(x_cutoff, color='cyan', linestyle='--', alpha=0.4)
for i, (cx, cy, r, interp) in enumerate(final):
    color = 'orange' if interp else 'red'
    # center dot
    ax.scatter(cx, cy, facecolors='none' if interp else color,
               edgecolors=color, s=80, linewidths=2, marker='o')
    # radius circle (the actual crop footprint)
    ax.add_patch(plt.Circle((cx, cy), r, fill=False,
                            edgecolor=color, linewidth=2, alpha=0.9))
    # optional: show the 0.94 crop the Cellpose cell actually uses
    ax.add_patch(plt.Circle((cx, cy), int(r * 0.94), fill=False,
                            edgecolor=color, linestyle=':', linewidth=1, alpha=0.6))
    ax.text(cx + 10, cy + 10, str(i+1), color='yellow', fontsize=14, fontweight='bold')
ax.set_aspect('equal')          # keep circles round, not oval
plt.show()

# --- 5. Bridge to the Cellpose cell: (x, y, r) triples ---
valid_wells = [(x, y, r) for (x, y, r, interp) in final]
```

## 3. Segment colonies per well (Cellpose)

For each detected well: crop to a circle, build a color-agnostic "distance from background" signal in LAB space, then run Cellpose to segment/count colonies. Shows raw crop vs. segmentation side by side.

```python
labels = [
    "Top_A1", "Top_A2", "Top_B1", "Top_B2", "Top_C1", "Top_C2",
    "Bottom_A1", "Bottom_A2", "Bottom_B1", "Bottom_B2", "Bottom_C1", "Bottom_C2"
]

report_data = []
print("Extracting wells and generating colony masks...\n")

for idx, (x, y, r) in enumerate(valid_wells):
    if idx >= len(labels):
        break
    label = labels[idx]

    # Crop from img_rgb (known-good color), NOT img
    crop_r = int(r * 0.94)
    y_min, y_max = max(0, y - crop_r), min(img_rgb.shape[0], y + crop_r)
    x_min, x_max = max(0, x - crop_r), min(img_rgb.shape[1], x + crop_r)

    well_crop = img_rgb[y_min:y_max, x_min:x_max].copy()   # RGB

    # Circular mask
    mask = np.zeros(well_crop.shape[:2], dtype="uint8")
    cv2.circle(mask, (well_crop.shape[1] // 2, well_crop.shape[0] // 2), crop_r, 255, -1)
    final_well = cv2.bitwise_and(well_crop, well_crop, mask=mask)   # RGB

    # --- Color-agnostic signal: distance from background in LAB ---
    lab = cv2.cvtColor(final_well, cv2.COLOR_RGB2LAB).astype(np.float32)   # RGB2LAB now

    ring = cv2.subtract(mask, cv2.erode(mask, np.ones((60, 60), np.uint8)))
    bg_a = np.median(lab[:, :, 1][ring > 0])
    bg_b = np.median(lab[:, :, 2][ring > 0])

    dist = np.sqrt((lab[:, :, 1] - bg_a) ** 2 + (lab[:, :, 2] - bg_b) ** 2)
    signal = cv2.normalize(dist, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
    signal = cv2.bitwise_and(signal, signal, mask=mask)

    # --- Cellpose ---
    masks_cp, flows, styles = cp_model.eval(
        signal,
        diameter=40,
        flow_threshold=0.9,
        cellprob_threshold=-2.0,
        normalize={"normalize": True, "percentile": [1.0, 99.0]},
    )

    colony_count = int(masks_cp.max())
    report_data.append({
        "Plate": input_path.split('/')[-1],
        "Well": label,
        "Colonies": colony_count
    })

    # --- Visualization ---
    fig, axes = plt.subplots(1, 2)
    axes[0].imshow(final_well)          # already RGB — no conversion
    axes[0].set_title(f"Raw Well: {label}")
    axes[0].axis('off')

    axes[1].imshow(signal, cmap='magma')
    axes[1].imshow(masks_cp, cmap='nipy_spectral', alpha=0.4)
    axes[1].set_title(f"AI Count: {colony_count} Colonies")
    axes[1].axis('off')
    plt.show()
```

## 4. Save results

Collects per-well colony counts into a DataFrame and writes `<plate_name>_Results.csv`.

```python
# Convert our dictionary to a clean Pandas DataFrame
df = pd.DataFrame(report_data)

# Display the spreadsheet directly in the notebook for a final review
display(df)

# Save to disk
base_name = os.path.splitext(os.path.basename(input_path))[0]
csv_name = f"{base_name}_Results.csv"
df.to_csv(csv_name, index=False)

print(f"💾 Data successfully saved to {csv_name}")
```
