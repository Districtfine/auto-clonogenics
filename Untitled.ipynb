# Colony counting pipeline

Isolate each well from a multi-well plate `.tif` scan, then run each well crop through Cellpose to count colonies and estimate diameter. Source scans live in `../Clonogenics`.

```python
import os
import glob
import warnings
warnings.filterwarnings("ignore", message="Sparse invariant checks")

import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tifffile as tifi
import torch
from ultralytics import FastSAM
from cellpose import models

# Make our inline plots look nice and big
plt.rcParams['figure.figsize'] = [12, 6]
```

## 1. Pick device & load models

Picks `mps`/`cuda`/`cpu` automatically (portable between Apple Silicon and the 1650 Ti box), loads FastSAM (well extractor) and Cellpose (colony counter) once. Reused by both single-image and batch runs below.

```python
# Portable device pick: mps on Apple Silicon, cuda on the 1650 Ti box, cpu fallback.
# DEVICE (str) feeds ultralytics/FastSAM; TORCH_DEVICE (torch.device) feeds Cellpose.
if torch.backends.mps.is_available():
    DEVICE = "mps"
elif torch.cuda.is_available():
    DEVICE = "cuda"
else:
    DEVICE = "cpu"
TORCH_DEVICE = torch.device(DEVICE)
print(f"Using device: {DEVICE}")

print("Loading FastSAM (Well Extractor) into GPU...")
sam_model = FastSAM('FastSAM-s.pt')

print("Loading Cellpose (Colony Counter) into GPU...")
cp_model = models.CellposeModel(gpu=DEVICE != "cpu", device=TORCH_DEVICE, model_type='cpsam_v2')
print("Done!")
```

## 2. Detect plate boundary & compute wells analytically

No per-well detection — just find each plate's outer rectangle, then place wells by arithmetic:

1. **Cyan cutoff** at `X_LIMIT_FRAC` truncates the empty right-hand plates + handwriting (drawn on the diagnostic so you can slide it).
2. `detect_plate_rects` — Otsu **threshold** (plate vs background is near-binary) → morphology to fill each plate solid → largest boxy contours → `boundingRect`. Keeps the top `N_PLATES`, sorts top→bottom (A, B). More stable than circle/edge fitting. Flip `THRESH_INVERT` if the box grabs the background instead of the plate. (Plates read near axis-aligned; if a rig adds rotation, swap in `minAreaRect` + a warp here.)
3. `detect_wells` — well centers are the **known fractional layout** of a standard 6-well plate (`WELL_X_FRACS`, `WELL_Y_FRACS`) applied to the plate box; radius is `WELL_R_FRAC` of plate width. Calibrate those fractions once from a clean scan and reuse for every plate from the same rig. Row-major labels `A1`...`A6`, `B1`...`B6`.

Diagnostic overlay: cyan cutoff, lime plate boxes, red well positions. Tune `debug_plate_edges.py` if the threshold misbehaves.

```python
X_LIMIT_FRAC = 0.55  # cyan cutoff line: keep only the plates left of this fraction of width.
N_PLATES = 2         # two stacked 6-well plates (left side of the scan)
PLATE_LETTERS = ["A", "B"]  # A = top plate, B = bottom plate

# --- Plate-boundary detection ---
# Threshold can't separate these transparent plates from the similar-brightness lightbox
# background, so the boundary comes from the frame EDGE: CLAHE contrast-boost -> Canny ->
# close -> largest boxy contour. That reliably finds each plate's outer rectangle.
CLAHE_CLIP = 3.0
CANNY_LOW, CANNY_HIGH = 30, 90
PLATE_AREA_FRAC = (0.03, 0.45)  # a plate occupies this fraction of the (cropped) scan
PLATE_ASPECT    = (0.4, 2.2)    # bbox width/height range (permissive; scans vary)
PLATE_RECTANGULARITY = 0.6      # contour_area / bbox_area — how box-like the shape must be

# --- Analytic 6-well layout: well centers as fractions of the plate bounding box ---
# Calibrated once from a clean scan; reuse for every plate from the same rig/camera.
# Standard 6-well plate geometry (39.1 mm pitch, 127.8 x 85.5 mm) gives these fractions.
WELL_X_FRACS = [0.27, 0.73]          # column centers  (fraction of plate width, left->right)
WELL_Y_FRACS = [0.19, 0.50, 0.81]    # row centers     (fraction of plate height, top->bottom)
WELL_R_FRAC  = 0.20                  # well radius as fraction of plate width


def detect_plate_rects(img_rgb, crop_x):
    """Find plate outer rectangles left of crop_x from their frame edges.

    Returns up to N_PLATES (x, y, w, h) boxes in full-image coords, sorted top-to-bottom.
    (Plates read near axis-aligned; if a rig adds rotation, swap boundingRect for
    cv2.minAreaRect + a warpAffine rectification here before computing well centers.)
    """
    gray = cv2.cvtColor(img_rgb[:, :crop_x], cv2.COLOR_RGB2GRAY)  # only look left of the cyan line
    # boost local contrast so the faint transparent-plastic borders survive Canny
    gray = cv2.createCLAHE(clipLimit=CLAHE_CLIP, tileGridSize=(8, 8)).apply(gray)
    blur = cv2.GaussianBlur(gray, (7, 7), 0)
    edges = cv2.Canny(blur, CANNY_LOW, CANNY_HIGH)
    # close gaps in the border so each plate is one solid contour
    edges = cv2.morphologyEx(edges, cv2.MORPH_CLOSE, np.ones((25, 25), np.uint8))

    contours, _ = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    height, width = gray.shape
    img_area = height * width

    candidates = []
    for cnt in contours:
        area = cv2.contourArea(cnt)
        if not (img_area * PLATE_AREA_FRAC[0] <= area <= img_area * PLATE_AREA_FRAC[1]):
            continue
        rect_x, rect_y, rect_w, rect_h = cv2.boundingRect(cnt)
        aspect = rect_w / rect_h if rect_h else 0
        if not (PLATE_ASPECT[0] <= aspect <= PLATE_ASPECT[1]):
            continue
        if area / (rect_w * rect_h) < PLATE_RECTANGULARITY:  # reject non-boxy blobs
            continue
        candidates.append((area, rect_x, rect_y, rect_w, rect_h))

    if not candidates:
        return []

    candidates.sort(reverse=True)  # by area desc — plates are the largest boxy shapes
    boxes = [(rect_x, rect_y, rect_w, rect_h) for _, rect_x, rect_y, rect_w, rect_h in candidates[:N_PLATES]]
    boxes.sort(key=lambda box: box[1] + box[3] / 2)  # top-to-bottom
    return boxes


def detect_wells(img_rgb, show_plot=True, save_path=None):
    """Compute well centers analytically from each detected plate box (no per-well detection).

    Labels A1-A6 (top plate), B1-B6 (bottom plate). Returns (x, y, r, label) tuples.
    """
    crop_x = int(img_rgb.shape[1] * X_LIMIT_FRAC)
    plates = detect_plate_rects(img_rgb, crop_x)

    final = []      # (x, y, r, label)
    for plate_idx, (px, py, pw, ph) in enumerate(plates):
        plate_letter = PLATE_LETTERS[plate_idx] if plate_idx < len(PLATE_LETTERS) else str(plate_idx)
        well_r = int(pw * WELL_R_FRAC)

        # row-major over the known fractional layout: A1=top-left, A2=top-right, A3=mid-left, ...
        for row_idx, y_frac in enumerate(WELL_Y_FRACS):
            for col_idx, x_frac in enumerate(WELL_X_FRACS):
                cx = int(px + pw * x_frac)
                cy = int(py + ph * y_frac)
                label = f"{plate_letter}{row_idx * len(WELL_X_FRACS) + col_idx + 1}"
                final.append((cx, cy, well_r, label))

    print(f"Detected {len(plates)}/{N_PLATES} plates -> {len(final)} wells.")

    if show_plot or save_path:
        fig, ax = plt.subplots(figsize=(10, 10))
        ax.imshow(img_rgb)
        ax.axvline(crop_x, color='cyan', linestyle='--', alpha=0.6)   # crop / cutoff line
        for (px, py, pw, ph) in plates:      # detected plate boxes in lime
            ax.add_patch(plt.Rectangle((px, py), pw, ph, fill=False,
                                       edgecolor='lime', linewidth=2, alpha=0.9))
        for (cx, cy, r, label) in final:     # analytic well positions in red
            ax.scatter(cx, cy, facecolors='red', edgecolors='red', s=80, marker='o')
            ax.add_patch(plt.Circle((cx, cy), r, fill=False, edgecolor='red', linewidth=2, alpha=0.9))
            ax.add_patch(plt.Circle((cx, cy), int(r * 0.94), fill=False,
                                    edgecolor='red', linestyle=':', linewidth=1, alpha=0.6))
            ax.text(cx + 10, cy + 10, label, color='yellow', fontsize=14, fontweight='bold')
        ax.set_aspect('equal')

        if save_path:
            os.makedirs(os.path.dirname(save_path), exist_ok=True)
            fig.savefig(save_path, dpi=150, bbox_inches='tight')
        if show_plot:
            plt.show()
        else:
            plt.close(fig)

    return final
```

## 3. Segment colonies per well (Cellpose)

Defines `count_colonies(img_rgb, valid_wells, plate_name, show_plots=True)`: for each detected well, crops to a circle, builds a color-agnostic "distance from background" signal in LAB space, then runs Cellpose to segment/count colonies. Returns a list of `{"Plate", "Well", "Colonies"}` rows. Reusable per-image, same as `detect_wells`.

```python
def count_colonies(img_rgb, valid_wells, plate_name, show_plots=True, save_dir=None):
    """Segment colonies in each detected well and return per-well colony counts.

    show_plots displays each well's raw/segmentation panel inline; save_dir (if given)
    writes one PNG per well to disk instead/as well.
    """
    report_data = []
    print(f"Extracting wells and generating colony masks for {plate_name}...\n")

    if save_dir:
        os.makedirs(save_dir, exist_ok=True)

    for (x, y, r, label) in valid_wells:
        # Crop from img_rgb (known-good color), NOT the raw img
        crop_r = int(r * 0.94)
        y_min, y_max = max(0, y - crop_r), min(img_rgb.shape[0], y + crop_r)
        x_min, x_max = max(0, x - crop_r), min(img_rgb.shape[1], x + crop_r)

        well_crop = img_rgb[y_min:y_max, x_min:x_max].copy()   # RGB

        # Circular mask
        mask = np.zeros(well_crop.shape[:2], dtype="uint8")
        cv2.circle(mask, (well_crop.shape[1] // 2, well_crop.shape[0] // 2), crop_r, 255, -1)
        final_well = cv2.bitwise_and(well_crop, well_crop, mask=mask)   # RGB

        # --- Color-agnostic signal: distance from background in LAB ---
        lab = cv2.cvtColor(final_well, cv2.COLOR_RGB2LAB).astype(np.float32)

        ring = cv2.subtract(mask, cv2.erode(mask, np.ones((60, 60), np.uint8)))
        bg_a = np.median(lab[:, :, 1][ring > 0])
        bg_b = np.median(lab[:, :, 2][ring > 0])

        dist = np.sqrt((lab[:, :, 1] - bg_a) ** 2 + (lab[:, :, 2] - bg_b) ** 2)
        signal = cv2.normalize(dist, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
        signal = cv2.bitwise_and(signal, signal, mask=mask)

        # --- Cellpose ---
        masks_cp, flows, styles = cp_model.eval(
            signal,
            diameter=40,
            flow_threshold=0.9,
            cellprob_threshold=-2.0,
            normalize={"normalize": True, "percentile": [1.0, 99.0]},
        )

        colony_count = int(masks_cp.max())
        report_data.append({
            "Plate": plate_name,
            "Well": label,
            "Colonies": colony_count
        })

        if show_plots or save_dir:
            fig, axes = plt.subplots(1, 2)
            axes[0].imshow(final_well)          # already RGB — no conversion
            axes[0].set_title(f"Raw Well: {label}")
            axes[0].axis('off')

            axes[1].imshow(signal, cmap='magma')
            axes[1].imshow(masks_cp, cmap='nipy_spectral', alpha=0.4)
            axes[1].set_title(f"AI Count: {colony_count} Colonies")
            axes[1].axis('off')

            if save_dir:
                well_path = os.path.join(save_dir, f"{label}.png")
                fig.savefig(well_path, dpi=150, bbox_inches='tight')
            if show_plots:
                plt.show()
            else:
                plt.close(fig)

    return report_data
```

## 4. Choose mode & run

`BATCH_MODE = False` runs a single file; `True` sweeps every `.tif` in `CLONOGENICS_DIR`. `WELLS_ONLY = True` stops after well-detection (skips the slower Cellpose step) — useful for checking/tuning well detection across a whole folder without waiting on colony segmentation. In batch mode, plots always write to `BATCH_OUTPUT_DIR/<plate_name>/` instead of showing inline; single-image mode always shows inline. Toggle the flags and re-run this + the save cell below — nothing else needs to change.

```python
BATCH_MODE = True   # single image while tuning plate detection; flip to True to sweep the folder
WELLS_ONLY = True    # True: stop after well-detection, skip Cellpose

# Single-image mode: which file to run
input_path = "../Clonogenics/4h HS 43 + radiation001.tif"

# Batch mode: which folder to sweep, and where to write images (instead of inline display)
CLONOGENICS_DIR = "../Clonogenics"
BATCH_OUTPUT_DIR = "batch_output"


def process_plate(path, show_plots, save_dir=None, wells_only=False):
    plate_name = os.path.basename(path)
    print(f"Reading {plate_name}...")
    img = tifi.imread(path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_GRAY2RGB) if len(img.shape) == 2 else img[..., :3]

    # detect_wells crops to left of the cyan cutoff (X_LIMIT_FRAC) internally, so pass the
    # full image; well coords come back in full-image space for count_colonies to reuse.
    grid_path = os.path.join(save_dir, plate_name, "grid.png") if save_dir else None
    wells_dir = os.path.join(save_dir, plate_name) if save_dir else None

    valid_wells = detect_wells(img_rgb, show_plot=show_plots, save_path=grid_path)

    if wells_only:
        return [{"Plate": plate_name, "Well": label, "x": x, "y": y, "r": r}
                for (x, y, r, label) in valid_wells]

    return count_colonies(img_rgb, valid_wells, plate_name, show_plots=show_plots, save_dir=wells_dir)


if BATCH_MODE:
    tif_paths = sorted(glob.glob(os.path.join(CLONOGENICS_DIR, "*.tif")))
    print(f"Batch mode: {len(tif_paths)} plates found in {CLONOGENICS_DIR}")
    print(f"Writing well/grid images to {BATCH_OUTPUT_DIR}/<plate_name>/\n")
    report_data = []
    for path in tif_paths:
        print(f"\n=== {os.path.basename(path)} ===")
        # WELLS_ONLY is a diagnostic step — show inline even in batch
        report_data.extend(process_plate(
            path, show_plots=WELLS_ONLY, save_dir=BATCH_OUTPUT_DIR, wells_only=WELLS_ONLY
        ))
else:
    report_data = process_plate(input_path, show_plots=True, wells_only=WELLS_ONLY)
```

```python
# Convert results to a clean Pandas DataFrame
df = pd.DataFrame(report_data)

# Display the spreadsheet directly in the notebook for a final review
display(df)

# Save to disk: one combined CSV for batch runs, per-plate CSV for single-image runs;
# "wells" vs "Results" in the name so a wells-only run doesn't overwrite a full run's CSV
suffix = "wells" if WELLS_ONLY else "Results"
if BATCH_MODE:
    csv_name = f"batch_{suffix}.csv"
else:
    csv_name = f"{os.path.splitext(os.path.basename(input_path))[0]}_{suffix}.csv"
df.to_csv(csv_name, index=False)

print(f"💾 Data successfully saved to {csv_name}")
```
