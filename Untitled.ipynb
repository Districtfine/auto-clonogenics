# Colony counting pipeline

Isolate each well from a multi-well plate `.tif` scan, then run each well crop through Cellpose to count colonies and estimate diameter. Source scans live in `../Clonogenics`.

```python
import os
import glob
import warnings
warnings.filterwarnings("ignore", message="Sparse invariant checks")

import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tifffile as tifi
import torch
from ultralytics import FastSAM
from cellpose import models

# Make our inline plots look nice and big
plt.rcParams['figure.figsize'] = [12, 6]
```

## 1. Pick device & load models

Picks `mps`/`cuda`/`cpu` automatically (portable between Apple Silicon and the 1650 Ti box), loads FastSAM (well extractor) and Cellpose (colony counter) once. Reused by both single-image and batch runs below.

```python
# Portable device pick: mps on Apple Silicon, cuda on the 1650 Ti box, cpu fallback.
# DEVICE (str) feeds ultralytics/FastSAM; TORCH_DEVICE (torch.device) feeds Cellpose.
if torch.backends.mps.is_available():
    DEVICE = "mps"
elif torch.cuda.is_available():
    DEVICE = "cuda"
else:
    DEVICE = "cpu"
TORCH_DEVICE = torch.device(DEVICE)
print(f"Using device: {DEVICE}")

print("Loading FastSAM (Well Extractor) into GPU...")
sam_model = FastSAM('FastSAM-s.pt')

print("Loading Cellpose (Colony Counter) into GPU...")
cp_model = models.CellposeModel(gpu=DEVICE != "cpu", device=TORCH_DEVICE, model_type='cpsam_v2')
print("Done!")
```

## 2. Detect wells & build grid

Defines `detect_wells(img_rgb, show_plot=True)`: runs FastSAM on a plate image, filters detections by circularity/area to find stained wells, then snaps them onto a consensus 2-plate × (2 col × 3 row) grid so missed/occluded wells still get an inferred position. Returns `valid_wells`: `(x, y, r, label)` per well, e.g. `A1`...`A6`, `B1`...`B6`. Reusable per-image — called once for single-image mode, once per file in batch mode.

```python
X_LIMIT_FRAC = 0.60
MIN_SPACING  = 150
N_COLS = 2
N_ROWS_PER_PLATE = 3
N_PLATES = 2  # two stacked 6-well plates
PLATE_LETTERS = ["A", "B"]  # A = top plate, B = bottom plate


def split_into_k(values, k):
    vals = sorted(values)
    if len(vals) <= k:
        return [int(v) for v in vals]
    gaps = sorted(((vals[i+1]-vals[i], i) for i in range(len(vals)-1)), reverse=True)
    cut_after = sorted(i for _, i in gaps[:k-1])
    groups, start = [], 0
    for c in cut_after:
        groups.append(vals[start:c+1]); start = c+1
    groups.append(vals[start:])
    return [int(np.mean(g)) for g in groups]


def pad_to_k(values, k, gap):
    """Extend a sorted list of grid-line positions up to length k using a known spacing.

    Needed when a plate has fewer detected wells than N_COLS/N_ROWS_PER_PLATE in some
    axis, so split_into_k returns fewer than k groups. Extra positions are extrapolated
    and end up flagged as inferred (same handling as any other missing well).
    """
    values = sorted(values)
    if len(values) >= k:
        return values[:k]
    if not values:
        return [0] * k
    gap = gap if gap else 1
    padded = list(values)
    while len(padded) < k:
        padded.append(padded[-1] + gap)
    return padded


def group_by_gaps(points, axis, k):
    """Partition points into k groups along an axis (0=x, 1=y) at the k-1 largest gaps."""
    pts = sorted(points, key=lambda p: p[axis])
    if len(pts) <= k:
        return [[p] for p in pts]
    coords = [p[axis] for p in pts]
    gaps = sorted(((coords[i+1]-coords[i], i) for i in range(len(coords)-1)), reverse=True)
    cut_after = sorted(i for _, i in gaps[:k-1])
    groups, start = [], 0
    for c in cut_after:
        groups.append(pts[start:c+1]); start = c+1
    groups.append(pts[start:])
    return groups


def detect_wells(img_rgb, show_plot=True, save_path=None):
    """Detect stained wells in a plate image and label them A1-A6 (top plate), B1-B6 (bottom plate).

    show_plot displays the grid preview inline; save_path (if given) writes it to disk instead/as well.
    Returns a list of (x, y, r, label) tuples.
    """
    min_well_area = img_rgb.shape[0] * img_rgb.shape[1] * 0.008  # well ~ 0.8%+ of scan area
    x_cutoff = img_rgb.shape[1] * X_LIMIT_FRAC

    results = sam_model.predict(source=img_rgb, device=DEVICE, imgsz=1024, conf=0.25, iou=0.5, verbose=False)
    masks = results[0].masks

    # --- 1. Detect stained wells ---
    detected = []
    for mask in masks.xy:
        cnt = np.array(mask, dtype=np.int32)
        area = cv2.contourArea(cnt)
        perimeter = cv2.arcLength(cnt, True)
        circularity = 4*np.pi*area/(perimeter**2) if perimeter > 0 else 0
        if area > min_well_area and circularity > 0.7:
            M = cv2.moments(cnt)
            if M["m00"] != 0:
                cx, cy = int(M["m10"]/M["m00"]), int(M["m01"]/M["m00"])
                if cx <= x_cutoff and all((cx-ox)**2+(cy-oy)**2 >= MIN_SPACING**2
                                          for ox, oy in detected):
                    detected.append((cx, cy))

    # --- 2. Split detections into the two physical plates, then grid each one ---
    all_col_xs = split_into_k([p[0] for p in detected], N_COLS)
    col_gap = np.mean(np.diff(sorted(all_col_xs))) if len(all_col_xs) > 1 else 0

    # split detections into the two plates by the large vertical gap between them,
    # then explicitly sort plates top-to-bottom by mean y (so plate 0 = A = top)
    plates = group_by_gaps(detected, axis=1, k=N_PLATES)
    plates = sorted(plates, key=lambda pts: np.mean([p[1] for p in pts]))

    final = []
    for plate_idx, plate_pts in enumerate(plates):
        plate_letter = PLATE_LETTERS[plate_idx] if plate_idx < len(PLATE_LETTERS) else str(plate_idx)

        col_xs = split_into_k([p[0] for p in plate_pts], N_COLS)
        row_ys = split_into_k([p[1] for p in plate_pts], N_ROWS_PER_PLATE)

        row_gap = np.mean(np.diff(sorted(row_ys))) if len(row_ys) > 1 else col_gap
        well_r  = int(min(g for g in (col_gap, row_gap) if g > 0) / 2)

        # a plate with fewer detected wells than N_COLS/N_ROWS_PER_PLATE in some axis
        # leaves col_xs/row_ys short — pad using the known spacing before indexing into them
        col_xs = pad_to_k(col_xs, N_COLS, col_gap)
        row_ys = pad_to_k(row_ys, N_ROWS_PER_PLATE, row_gap)

        # assign each detected well in THIS plate to its nearest (row, col) cell
        assign = {}
        for (px, py) in plate_pts:
            c = min(range(len(col_xs)), key=lambda c: abs(px - col_xs[c]))
            r = min(range(len(row_ys)), key=lambda r: abs(py - row_ys[r]))
            assign[(r, c)] = (px, py)

        # consensus grid lines from the detections (median => robust to one bad centroid)
        def col_consensus(c):
            xs = [pos[0] for (rr, cc), pos in assign.items() if cc == c]
            return np.median(xs) if xs else col_xs[c]
        def row_consensus(r):
            ys = [pos[1] for (rr, cc), pos in assign.items() if rr == r]
            return np.median(ys) if ys else row_ys[r]

        tol_x = 0.4 * col_gap          # a detection this far off its column is an outlier
        tol_y = 0.4 * row_gap

        # r = row (top to bottom), c = col (left to right); row-major so numbering
        # goes across each row before moving down: e.g. A1=top-left, A2=top-right,
        # A3=middle-left, A4=middle-right, A5=bottom-left, A6=bottom-right.
        for r in range(N_ROWS_PER_PLATE):
            for c in range(N_COLS):
                well_label = f"{plate_letter}{r * N_COLS + c + 1}"
                exp_x, exp_y = int(col_consensus(c)), int(row_consensus(r))
                if (r, c) in assign:
                    px, py = assign[(r, c)]
                    x = exp_x if abs(px - exp_x) > tol_x else px   # snap only gross outliers
                    y = exp_y if abs(py - exp_y) > tol_y else py
                    snapped = (x != px) or (y != py)
                    final.append((x, y, well_r, snapped, well_label))  # snapped => flagged like inferred
                else:
                    final.append((exp_x, exp_y, well_r, True, well_label))

    print(f"Final Count: {len(final)} wells "
          f"({sum(not f[3] for f in final)} detected, {sum(f[3] for f in final)} inferred/snapped).")

    if show_plot or save_path:
        fig, ax = plt.subplots(figsize=(10, 10))
        ax.imshow(img_rgb)
        ax.axvline(x_cutoff, color='cyan', linestyle='--', alpha=0.4)
        for (cx, cy, r, interp, well_label) in final:
            color = 'orange' if interp else 'red'
            ax.scatter(cx, cy, facecolors='none' if interp else color,
                       edgecolors=color, s=80, linewidths=2, marker='o')
            ax.add_patch(plt.Circle((cx, cy), r, fill=False,
                                    edgecolor=color, linewidth=2, alpha=0.9))
            # optional: show the 0.94 crop the Cellpose cell actually uses
            ax.add_patch(plt.Circle((cx, cy), int(r * 0.94), fill=False,
                                    edgecolor=color, linestyle=':', linewidth=1, alpha=0.6))
            ax.text(cx + 10, cy + 10, well_label, color='yellow', fontsize=14, fontweight='bold')
        ax.set_aspect('equal')          # keep circles round, not oval

        if save_path:
            os.makedirs(os.path.dirname(save_path), exist_ok=True)
            fig.savefig(save_path, dpi=150, bbox_inches='tight')
        if show_plot:
            plt.show()
        else:
            plt.close(fig)

    return [(x, y, well_r, well_label) for (x, y, well_r, interp, well_label) in final]
```

## 3. Segment colonies per well (Cellpose)

Defines `count_colonies(img_rgb, valid_wells, plate_name, show_plots=True)`: for each detected well, crops to a circle, builds a color-agnostic "distance from background" signal in LAB space, then runs Cellpose to segment/count colonies. Returns a list of `{"Plate", "Well", "Colonies"}` rows. Reusable per-image, same as `detect_wells`.

```python
def count_colonies(img_rgb, valid_wells, plate_name, show_plots=True, save_dir=None):
    """Segment colonies in each detected well and return per-well colony counts.

    show_plots displays each well's raw/segmentation panel inline; save_dir (if given)
    writes one PNG per well to disk instead/as well.
    """
    report_data = []
    print(f"Extracting wells and generating colony masks for {plate_name}...\n")

    if save_dir:
        os.makedirs(save_dir, exist_ok=True)

    for (x, y, r, label) in valid_wells:
        # Crop from img_rgb (known-good color), NOT the raw img
        crop_r = int(r * 0.94)
        y_min, y_max = max(0, y - crop_r), min(img_rgb.shape[0], y + crop_r)
        x_min, x_max = max(0, x - crop_r), min(img_rgb.shape[1], x + crop_r)

        well_crop = img_rgb[y_min:y_max, x_min:x_max].copy()   # RGB

        # Circular mask
        mask = np.zeros(well_crop.shape[:2], dtype="uint8")
        cv2.circle(mask, (well_crop.shape[1] // 2, well_crop.shape[0] // 2), crop_r, 255, -1)
        final_well = cv2.bitwise_and(well_crop, well_crop, mask=mask)   # RGB

        # --- Color-agnostic signal: distance from background in LAB ---
        lab = cv2.cvtColor(final_well, cv2.COLOR_RGB2LAB).astype(np.float32)

        ring = cv2.subtract(mask, cv2.erode(mask, np.ones((60, 60), np.uint8)))
        bg_a = np.median(lab[:, :, 1][ring > 0])
        bg_b = np.median(lab[:, :, 2][ring > 0])

        dist = np.sqrt((lab[:, :, 1] - bg_a) ** 2 + (lab[:, :, 2] - bg_b) ** 2)
        signal = cv2.normalize(dist, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
        signal = cv2.bitwise_and(signal, signal, mask=mask)

        # --- Cellpose ---
        masks_cp, flows, styles = cp_model.eval(
            signal,
            diameter=40,
            flow_threshold=0.9,
            cellprob_threshold=-2.0,
            normalize={"normalize": True, "percentile": [1.0, 99.0]},
        )

        colony_count = int(masks_cp.max())
        report_data.append({
            "Plate": plate_name,
            "Well": label,
            "Colonies": colony_count
        })

        if show_plots or save_dir:
            fig, axes = plt.subplots(1, 2)
            axes[0].imshow(final_well)          # already RGB — no conversion
            axes[0].set_title(f"Raw Well: {label}")
            axes[0].axis('off')

            axes[1].imshow(signal, cmap='magma')
            axes[1].imshow(masks_cp, cmap='nipy_spectral', alpha=0.4)
            axes[1].set_title(f"AI Count: {colony_count} Colonies")
            axes[1].axis('off')

            if save_dir:
                well_path = os.path.join(save_dir, f"{label}.png")
                fig.savefig(well_path, dpi=150, bbox_inches='tight')
            if show_plots:
                plt.show()
            else:
                plt.close(fig)

    return report_data
```

## 4. Choose mode & run

`BATCH_MODE = False` runs a single file; `True` sweeps every `.tif` in `CLONOGENICS_DIR`. `WELLS_ONLY = True` stops after well-detection (skips the slower Cellpose step) — useful for checking/tuning well detection across a whole folder without waiting on colony segmentation. In batch mode, plots always write to `BATCH_OUTPUT_DIR/<plate_name>/` instead of showing inline; single-image mode always shows inline. Toggle the flags and re-run this + the save cell below — nothing else needs to change.

```python
BATCH_MODE = True
WELLS_ONLY = True  # True: stop after well-detection, skip Cellpose

# Single-image mode: which file to run
input_path = "../Clonogenics/4h HS 43 + radiation001.tif"

# Batch mode: which folder to sweep, and where to write images (instead of inline display)
CLONOGENICS_DIR = "../Clonogenics"
BATCH_OUTPUT_DIR = "batch_output"


def process_plate(path, show_plots, save_dir=None, wells_only=False):
    plate_name = os.path.basename(path)
    print(f"Reading {plate_name}...")
    img = tifi.imread(path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_GRAY2RGB) if len(img.shape) == 2 else img[..., :3]

    grid_path = os.path.join(save_dir, plate_name, "grid.png") if save_dir else None
    wells_dir = os.path.join(save_dir, plate_name) if save_dir else None

    valid_wells = detect_wells(img_rgb, show_plot=show_plots, save_path=grid_path)

    if wells_only:
        return [{"Plate": plate_name, "Well": label, "x": x, "y": y, "r": r}
                for (x, y, r, label) in valid_wells]

    return count_colonies(img_rgb, valid_wells, plate_name, show_plots=show_plots, save_dir=wells_dir)


if BATCH_MODE:
    tif_paths = sorted(glob.glob(os.path.join(CLONOGENICS_DIR, "*.tif")))
    print(f"Batch mode: {len(tif_paths)} plates found in {CLONOGENICS_DIR}")
    print(f"Writing well/grid images to {BATCH_OUTPUT_DIR}/<plate_name>/\n")
    report_data = []
    for path in tif_paths:
        print(f"\n=== {os.path.basename(path)} ===")
        # WELLS_ONLY is a diagnostic step (FastSAM is cheap) — show inline even in batch
        report_data.extend(process_plate(
            path, show_plots=WELLS_ONLY, save_dir=BATCH_OUTPUT_DIR, wells_only=WELLS_ONLY
        ))
else:
    report_data = process_plate(input_path, show_plots=True, wells_only=WELLS_ONLY)
```

```python
# Convert results to a clean Pandas DataFrame
df = pd.DataFrame(report_data)

# Display the spreadsheet directly in the notebook for a final review
display(df)

# Save to disk: one combined CSV for batch runs, per-plate CSV for single-image runs;
# "wells" vs "Results" in the name so a wells-only run doesn't overwrite a full run's CSV
suffix = "wells" if WELLS_ONLY else "Results"
if BATCH_MODE:
    csv_name = f"batch_{suffix}.csv"
else:
    csv_name = f"{os.path.splitext(os.path.basename(input_path))[0]}_{suffix}.csv"
df.to_csv(csv_name, index=False)

print(f"💾 Data successfully saved to {csv_name}")
```
